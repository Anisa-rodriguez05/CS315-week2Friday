```python
import requests
import csv
from bs4 import BeautifulSoup
from seleniumbase import Driver
```

```python
url = "https://mydramalist.com/shows/newest"

Dramas = []

numPages = 2 

for page in range (1, numPages + 1):
    page_url = f"{url}?page{page}"
    response = requests.get(page_url)
    soup = BeautifulSoup(response.content, "html.parser")

    cards = soup.find_all("div", class_="box")

    for card in cards:
        #rating = card.find("span", class_="score").get_text(strip=True)
        title1 = card.find("h6", class_="text-primary title")
        title = title1.get_text(strip=True) if title1 else "N/A"

        rating1 = card.find("span", class_="score")
        rating = rating1.get_text(strip=True) if rating1 else "N/A"

        info1 = card.find_all("p")
        info = info1[-1].get_text(strip=True) if info1 else "N/A"

        drama = {
            "title": title, 
            "rating": rating,
            "info": info
        }

        Dramas.append(drama)


print(Dramas)


```
